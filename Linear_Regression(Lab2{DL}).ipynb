```python
import numpy as np
import matplotlib.pyplot as plt
import torch
```

```python
np.random.seed(232)
x=np.linspace(0,10,100)
noise=np.random.random(100)
y=3*x+5+noise
print(x)
print(noise)
print(y)
```

```python
plt.figure(figsize=(7,4))
plt.scatter(x,y)
plt.grid(True)
plt.show()
```

```python
w=0.0
b=0.0
def predict(x,w,b):
  return w*x+b
y_initial=predict(x,w,b)
plt.figure(figsize=(7,4))
plt.scatter(x,y)
plt.plot(x,y_initial,'r')
plt.grid(True)
plt.show()
```

```python
def mse(y_true,y_pred):
  return np.mean((y_true-y_pred)**2)
print(mse(y,y_initial))
```

```python
def gradients(x,y,w,b):
  n=len(x)
  yp=predict(x,w,b)
  dw=(-2/n)*np.sum(x*(y-yp))
  db=(-2/n)*np.sum(y-yp)
  return dw,db
```

```python
dw,db=gradients(x,y,w,b)
print(dw,db)
w-=0.01*dw
b-=0.01*db
print(w,b)
```

```python
w=0.0
b=0.0
lr=0.01
epochs=100
loss_history=[]
weight_history=[]
bias_history=[]
for epoch in range(epochs):
  dw,db=gradients(x,y,w,b)
  w-=lr*dw
  b-=lr*db
  y_pred = predict(x, w, b)
  loss_history.append(mse(y, y_pred))
  weight_history.append(w)
  bias_history.append(b)
print(loss_history)
print(weight_history)
```

```python
ep=1000
tw,tb=0.0,0.0
for i in range(ep+1):
  dw,db=gradients(x,y,tw,tb)
  tw-=0.01*dw
  tb-=0.01*db
plt.figure(figsize=(12,12))
plt.scatter(x,y)
plt.plot(x,y_initial,'r')
plt.plot(x,predict(x,tw,tb),'r')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(12,12))
plt.plot(loss_history)
plt.plot(weight_history)
plt.plot(bias_history)
plt.grid(True)
plt.title('Loss')
```

```python
x_tensor=torch.tensor(x,dtype=torch.float32)
y_tensor=torch.tensor(y,dtype=torch.float32)
w=torch.tensor(0.0)
b=torch.tensor(0.0)
print(x,y,w,b)
```

```python
losses=[];weights=[];biases=[]
for epoch in range(1000):
    yp=w*x_tensor+b
    dw=(-2/len(x))*torch.sum(x_tensor*(y_tensor-yp))
    db=(-2/len(x))*torch.sum(y_tensor-yp)
    w=w-0.01*dw
    b=b-0.01*db
    losses.append(torch.mean((y_tensor-(w*x_tensor+b))**2).item())
    weights.append(w.item());biases.append(b.item())
plt.scatter(x,y)
plt.plot(x,(w+x_tensor+b).numpy(),'r')
plt.grid(True)
plt.show()
```

```python
x_tensor = torch.tensor(x, dtype=torch.float32)
y_tensor = torch.tensor(y, dtype=torch.float32)

# Tell PyTorch to track gradients for w and b
w = torch.tensor(0.0, requires_grad=True)
b = torch.tensor(0.0, requires_grad=True)

print(x, y, w, b)
```

```python
losses=[]; weights=[]; biases=[]
for epoch in range(1000):
  # Changed * to + for the bias
  yp = w * x_tensor + b

  loss = ((y_tensor - yp)**2).mean()
  loss.backward()

  with torch.no_grad():
    w -= 0.01 * w.grad
    b -= 0.01 * b.grad

  # Fixed the typo: 'zer0_' to 'zero_'
  w.grad.zero_()
  b.grad.zero_()

  losses.append(loss.item())
  weights.append(w.item())
  biases.append(b.item())
```
